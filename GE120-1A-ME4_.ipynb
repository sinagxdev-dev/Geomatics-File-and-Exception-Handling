```python
import csv
import math
# Loop until valid file is found
while True:
    filename = input("Enter the text file name: ")
    
    # Check if filename has .csv extension
    if not filename.endswith('.csv'):
        filename = filename + '.csv'

    # Try to open the file
    try:
        dataset = []
        with open(filename, mode='r', newline='') as file:
            reader = csv.DictReader(file)
            for row in reader:
                dataset.append(row)
        print(f"Opening {filename}")
        break
    except FileNotFoundError:
        print(f"{filename} not found. Check your filename.")
        print()
```

```python
try:
    import pandas as pd
    print("pandas imported successfully")
except ImportError:
    %pip install pandas
    import pandas as pd
    print("pandas imported successfully after installation.")
```

```python
dataframe = pd.read_csv(filename)
print(dataframe.head())

Nmax = dataframe['N'].max()
Nmin = dataframe['N'].min()
Emax = dataframe['E'].max()
Emin = dataframe['E'].min()

print(f"Northings - Max: {Nmax}, Min: {Nmin}")
print(f"Eastings - Max: {Emax}, Min: {Emin}")
```

```python
Ndiff = Nmax - Nmin
Ediff = Emax - Emin
print(Ndiff)
print(Ediff)

if Ndiff > Ediff:
    scale = Ndiff/0.3
    newscale = (math.ceil(round((scale/100),2)))*100
    if newscale in [700, 900, 7000, 9000]:
        newscale += 100
    print(newscale)

elif Ndiff < Ediff:
    scale = Ediff/0.3
    newscale = (math.ceil(round((scale/100),2)))*100
    if newscale in [700, 900, 7000, 9000]:
        newscale += 100
    print(newscale)
```

```python
Ncenter = (Nmax + Nmin)/2
Ecenter = (Emax + Emin)/2
print(Ncenter, Ecenter)

dataframe['diff_frNcenter'] = dataframe['N'] - Ncenter
dataframe['diff_frEcenter'] = dataframe['E'] - Ecenter

# Add plotting columns in cm
dataframe['N_plot_cm'] = (dataframe['diff_frNcenter'] / newscale) * 100
dataframe['E_plot_cm'] = (dataframe['diff_frEcenter'] / newscale) * 100

print(dataframe.head())
```

```python
import os

# Get output filename
outputfile = input("Enter the name of the output filename: ")

# Check if filename has .csv extension
if not outputfile.endswith('.csv'):
    outputfile = outputfile + '.csv'

print(outputfile)

# Check if file exists and save
if os.path.exists(outputfile):
    print(f"WARNING: {outputfile} already exists and will be overwritten.")
    
    while True:
        overwrite = input("Overwrite? (y/n): ")
        if overwrite == 'y':
            dataframe.to_csv(outputfile, index=False)
            print("File written successfully")
            break
        elif overwrite == 'n':
            print("Saving file cancelled")
            break
else:
    dataframe.to_csv(outputfile, index=False)
    print("File written successfully")
```
